```python
# training hparams
n_epochs        = 200
lr              = 0.0001
min_delta       = 0.0001
loss            = 'mse'


global_hparams = {
    "lr"                : lr,
    "n_epochs"          : n_epochs,
    "scheduler"         : 'plateau',
    "scheduler_kwargs"  : {'mode': 'min', 'factor': 0.5, 'patience': 5},
    'min_delta'         : min_delta,
    'loss'              : loss,
    'patience'          : 20,
    }

config = ExperimentConfig(
    name='commuter_embeddingmodel',
    data=DataConfig(
        disease_name        = 'influenza',
        nuts_level          = 'nuts3',
        min_date            = '2006-05-15',
        max_date            = '2020-06-01',
        split_berlin        = False,
        include_population  = True

    ),
    timeseries=TimeSeriesConfig(
        sequence_length = 1,
        horizon_size    = 1,
        horizon_leadtime= 3,
        lags            = 1
    ),
    preprocessing=PreprocessingConfig(
        log_transform_target= True,
        add_time_features   = True,
        normalization_method='zscore'
    ),
    splits=SplitConfig(
        split_trainval  = '2018-06-01',
        split_valtest   = '2019-06-01'
    ),
    model= ['GATv2EmbeddingsModel'],
    global_hparams = global_hparams,
    train_hparams  = {'verbose':2},
    graphs         = ['commuter1','commuter2','commuter3','commuter4','identity_graph','boolean_neighbors_self','boolean_neighbors_nonself'],
    baseline = True
)
runner = ExperimentRunner()
runner.define(config).execute('commuter_embeddingmodel')
```
